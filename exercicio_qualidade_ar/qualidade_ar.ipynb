#Parte 07
##Análise temporal

```python
df['data'] = pd.to_datetime(df['data'])
df['mes'] = df['data'].dt.month
df.groupby('mes')['indice_qualidade_ar'].mean()
```

##Perguntas

1. Em quais meses a qualidade do ar tende a ser pior?<br>
A qualidade tende a ser pior em fevereiro, janeiro e dezembro

```python
df.groupby('mes')['indice_qualidade_ar'].mean().sort_values()
```

2. Existe algum padrão sazonal?<br>
Analisando os dados, ele tende a ser pior no verão, categorizado pelas altas temperaturas
